```python
import itertools
import pandas as pd

# Definición de los niveles
niveles = ['Baja', 'Media', 'Alta']

# Generación de la matriz base 3^2
matriz_base = list(itertools.product(niveles, niveles))

# Inclusión de las 3 réplicas
replicas = 3
matriz_con_replicas = matriz_base * replicas

# Creación del DataFrame y aleatorización completa de las 27 corridas
df_doe = pd.DataFrame(matriz_con_replicas, columns=['Compliance', 'Resistencia'])
df_doe_aleatorio = df_doe.sample(frac=1, random_state=42).reset_index(drop=True)
df_doe_aleatorio.index += 1
df_doe_aleatorio.index.name = 'Orden_Ejecucion'

print("Matriz Aleatorizada - Diseño 3^2 con 3 Réplicas (27 corridas):")
print(df_doe_aleatorio.head(10)) # Mostrando las primeras 10 para visualización rápida
```

```python
# Exportar:
df_doe_aleatorio.to_csv('matriz_experimento_ventilador.csv')
```
